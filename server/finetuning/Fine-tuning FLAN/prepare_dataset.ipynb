```python
import pandas as pd

def txt_to_csv(txt_filepath, csv_filepath, delimiter='\t'):
    """
    Converts a text file to a CSV file using Pandas.

    Args:
        txt_filepath (str): Path to the input text file.
        csv_filepath (str): Path to save the output CSV file.
        delimiter (str, optional): The delimiter used in the text file. Defaults to tab.
    """
    try:
        df = pd.read_csv(txt_filepath, delimiter=delimiter)
        df.to_csv(csv_filepath, index=False)
        print(f"Successfully converted {txt_filepath} to {csv_filepath}")
    except Exception as e:
        print(f"Error during conversion: {e}")

# Example usage:
txt_file = "rsi_qa.txt"
csv_file = "rsi_qa.csv"
txt_to_csv(txt_file, csv_file, delimiter=',')
```

```python
import pandas as pd
import glob

def clean_combine_csv(file_pattern, output_file):
    all_dfs = []
    
    for file in glob.glob(file_pattern):
        # Read CSV while stripping whitespace from headers
        df = pd.read_csv(file)
        df.columns = df.columns.str.strip()  # Clean column names
        
        # Handle multiple Answer columns (including whitespace variations)
        answer_cols = [col for col in df.columns if 'Answers' in col]
        if len(answer_cols) > 1:
            # Combine all Answer columns
            df['Answers'] = df[answer_cols].bfill(axis=1).iloc[:, 0]
        
        # Select only Question and (cleaned) Answer columns
        df = df[['Question', 'Answers']].copy()
        
        # Remove empty rows
        df = df.dropna(how='all')
        all_dfs.append(df)
    
    # Combine and shuffle
    combined = pd.concat(all_dfs, ignore_index=True)
    shuffled = combined.sample(frac=1).reset_index(drop=True)
    
    # Save to CSV
    shuffled.to_csv(output_file, index=False)
    print(f"Successfully combined {len(all_dfs)} files with {len(shuffled)} Q&A pairs")

# Usage
clean_combine_csv("*_qa.csv", "data.csv")
```
